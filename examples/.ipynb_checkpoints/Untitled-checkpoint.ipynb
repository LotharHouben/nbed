```python
def bin_time_series(data_3d, bin_factor=2):
    """
    Bins a 3D time-series array (N_frames, Y, X) along the frame axis (axis 0).
    
    Parameters:
        data_3d (np.ndarray): Input 3D array of shape (N, Y, X)
        bin_factor (int): Number of consecutive frames to average together
        
    Returns:
        np.ndarray: Binned array of shape (N // bin_factor, Y, X)
        int: Actual number of frames binned
    """
    if bin_factor <= 1:
        return data_3d, 1

    num_frames, frame_y, frame_x = data_3d.shape
    
    # Calculate how many full bins fit into the total frame count
    n_binned = num_frames // bin_factor
    crop_limit = n_binned * bin_factor
    
    # Crop leftover trailing frames that don't fill a complete bin
    cropped_data = data_3d[:crop_limit]
    
    # Reshape to (n_binned, bin_factor, Y, X) and average over axis 1
    binned_data = cropped_data.reshape(n_binned, bin_factor, frame_y, frame_x).mean(axis=1)
    
    return binned_data, bin_factor
```

```python

def create_diffraction_player(
    data_3d, 
    frametime_ms=100.0, 
    dose_per_frame=1.0, 
    settling_time_ms=0.0,
    initial_fps=10
):
    """
    Interactive diffraction frame player with dose, time, and beam settling offset tracking.

    Parameters:
        data_3d (np.ndarray): 3D array of shape (num_frames, frame_y, frame_x)
        frametime_ms (float): Frame acquisition duration in milliseconds
        dose_per_frame (float): Electron dose per frame in e-/A^2
        settling_time_ms (float): Beam settling / lag time in milliseconds
        initial_fps (int): Initial playback frame rate
    """
    num_frames, frame_y, frame_x = data_3d.shape
    
    # Calculate default settling offset frames
    initial_offset = int(np.round(settling_time_ms / frametime_ms)) if frametime_ms > 0 else 0

    def apply_contrast(data, clip_percentiles, gamma):
        p_low, p_high = clip_percentiles
        vmin, vmax = np.percentile(data, [p_low, p_high])
        if vmax == vmin:
            vmax = vmin + 1e-5
        norm = np.clip((data - vmin) / (vmax - vmin), 0.0, 1.0)
        return np.power(norm, gamma)

    plt.ioff()
    fig, ax = plt.subplots(figsize=(6.5, 5))
    fig.canvas.header_visible = False

    # Adjust top subplot spacing explicitly to accommodate multi-line title
    fig.subplots_adjust(top=0.82)  # Reserves top ~18% of figure area for title text
    
    initial_frame = data_3d[0]
    initial_processed = apply_contrast(initial_frame, (0, 100), 1.0)
    
    im = ax.imshow(initial_processed, cmap='viridis', origin="lower", vmin=0, vmax=1)
    title = ax.set_title("Initializing...")
    fig.tight_layout()

    # --- Playback & Trim Controls ---
    play_widget = widgets.Play(
        value=0, min=0, max=num_frames - 1, step=1,
        interval=int(1000 / initial_fps), disabled=False
    )
    
    slider_frame = widgets.IntSlider(
        value=0, min=0, max=num_frames - 1, step=1,
        description='Frame:', continuous_update=True,
        layout=widgets.Layout(width='38%')
    )
    widgets.jslink((play_widget, 'value'), (slider_frame, 'value'))

    # Trim Range & Settling Offset Text Boxes    
    tb_start_frame = widgets.BoundedIntText(
        value=0,
        min=0,
        max=num_frames - 1,
        step=1,
        description='Start:',
        layout=widgets.Layout(width='120px'),
        style={'description_width': '50px'}
    )
    
    tb_end_frame = widgets.BoundedIntText(
        value=num_frames - 1, min=0, max=num_frames - 1, step=1,
        description='End:', layout=widgets.Layout(width='120px'),
        style={'description_width': '50px'}
    )
    tb_offset_frame = widgets.BoundedIntText(
        value=initial_offset, min=0, max=num_frames - 1, step=1,
        description='t_0 Offset:', layout=widgets.Layout(width='150px'), 
        style={'description_width': '90px'}
    )



    # Speed & Averaging Sliders
    slider_fps = widgets.IntSlider(
        value=initial_fps, min=1, max=60, step=1,
        description='FPS:', continuous_update=False,
        layout=widgets.Layout(width='140px')
    )
    slider_avg = widgets.IntSlider(
        value=1, min=1, max=51, step=2,
        description='Avg Win:', continuous_update=False,
        layout=widgets.Layout(width='170px')
    )

    # Contrast Controls
    slider_clip = widgets.FloatRangeSlider(
        value=[1., 99.8], min=0, max=100, step=0.5,
        description='Clip %:', continuous_update=False,
        layout=widgets.Layout(width='250px')
    )
    slider_gamma = widgets.FloatSlider(
        value=0.6, min=0.1, max=3.0, step=0.05,
        description='Gamma:', continuous_update=False,
        layout=widgets.Layout(width='160px')
    )
    dropdown_cmap = widgets.Dropdown(
        options=['inferno', 'magma', 'viridis', 'plasma', 'cividis', 'gray','gray_r'],
        value='inferno', description='Cmap:',
        layout=widgets.Layout(width='140px')
    )
    btn_reset = widgets.Button(description='Reset Contrast', button_style='warning')

    # --- Callbacks ---
    def update_trim_bounds(_=None):
        start_val = tb_start_frame.value
        end_val = tb_end_frame.value

        if start_val >= end_val:
            start_val = max(0, end_val - 1)
            tb_start_frame.value = start_val

        play_widget.min = start_val
        play_widget.max = end_val
        slider_frame.min = start_val
        slider_frame.max = end_val

        if slider_frame.value < start_val:
            slider_frame.value = start_val
        elif slider_frame.value > end_val:
            slider_frame.value = end_val
            
        update_display()

    def update_display(change=None):
        idx = slider_frame.value
        win = slider_avg.value
        start_idx_trim = tb_start_frame.value
        offset_idx = tb_offset_frame.value

        if win <= 1:
            raw_frame = data_3d[idx]
        else:
            half_win = win // 2
            start_idx = max(0, idx - half_win)
            end_idx = min(num_frames, idx + half_win + 1)
            raw_frame = np.mean(data_3d[start_idx:end_idx], axis=0)

        processed = apply_contrast(raw_frame, slider_clip.value, slider_gamma.value)
        im.set_data(processed)
        im.set_cmap(dropdown_cmap.value)
        
        # Calculate elapsed time (t=0 at Start + Offset)
        active_frame = idx - start_idx_trim - offset_idx
        elapsed_sec = active_frame * (frametime_ms / 1000.0)
        
        # Calculate dose (0 dose during settling frames)
        effective_dose_frames = max(0, active_frame + 1)
        cum_dose = effective_dose_frames * dose_per_frame

        title_line1 = f"Frame: {idx} / {num_frames - 1}  [{tb_start_frame.value}:{tb_end_frame.value}] (Offset: {offset_idx})"
        title_line2 = f"Time: {elapsed_sec:.3f} s  |  Fluence: {cum_dose:.0f} e⁻/Å²"
        
        title.set_text(f"{title_line1}\n{title_line2}")
        fig.canvas.draw_idle()

    def update_fps(change):
        play_widget.interval = int(1000 / slider_fps.value)

    def reset_contrast(b):
        slider_clip.value = [0, 100]
        slider_gamma.value = 1.0

    # Bindings
    tb_start_frame.observe(update_trim_bounds, names='value')
    tb_end_frame.observe(update_trim_bounds, names='value')
    tb_offset_frame.observe(update_display, names='value')

    slider_frame.observe(update_display, names='value')
    slider_avg.observe(update_display, names='value')
    slider_clip.observe(update_display, names='value')
    slider_gamma.observe(update_display, names='value')
    dropdown_cmap.observe(update_display, names='value')
    slider_fps.observe(update_fps, names='value')
    btn_reset.on_click(reset_contrast)

    # Initial title update
    update_display()

    # --- Layout ---
    controls_trim = widgets.HBox([tb_start_frame, tb_end_frame, tb_offset_frame])
    controls_playback = widgets.HBox([play_widget, slider_frame, slider_fps, slider_avg])
    controls_display = widgets.HBox([slider_clip, slider_gamma, dropdown_cmap, btn_reset])

    if hasattr(fig.canvas, 'header_visible'):
        canvas_widget = fig.canvas
    else:
        out = widgets.Output()
        with out:
            display(fig)
        canvas_widget = out

    ui = widgets.VBox([canvas_widget, controls_trim, controls_playback, controls_display])
    plt.ion()
    display(ui)
```

```python
b, bfact=bin_time_series(a, bin_factor=10)
```

```python
create_diffraction_player(b, frametime_ms=1.0*bfact, settling_time_ms=50.0/bfact, dose_per_frame=0.259*bfact, initial_fps=10)
```
